```python
import pandas as pd
import yfinance as yf

df = pd.read_csv("news_impact/data/events.csv")
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

start = df["date"].min() - pd.Timedelta(days=30)
end = df["date"].max() + pd.Timedelta(days=5)

def get_close(ticker, name):
    data = yf.download(ticker, start=start, end=end, auto_adjust=False, progress=False)
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)
    s = data["Close"]
    s.name = name
    return s

dxy = get_close("DX-Y.NYB", "dxy")
vix = get_close("^VIX", "vix")
gold = get_close("GC=F", "gold_daily")

market = pd.concat([dxy, vix, gold], axis=1)
market.index = pd.to_datetime(market.index)
```

```python
import numpy as np
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# merge market data onto event dates (forward-fill for weekends/holidays)
full_range = pd.date_range(market.index.min(), market.index.max(), freq="D")
market = market.reindex(full_range).ffill()

df = df.merge(market, left_on="date", right_index=True, how="left")
# build features
df["dxy_direction"] = np.sign(df["dxy"].diff())
df["vix_level"] = df["vix"]
df["gold_momentum_5d"] = df["gold_daily"].pct_change(5)

# time since last HIGH-impact event
last_high = None
gaps = []
for date, label in zip(df["date"], df["impact_label"]):
    gaps.append((date - last_high).days if last_high else np.nan)
    if label == "HIGH":
        last_high = date
df["time_since_last_high"] = gaps

# scale + impute
feature_cols = ["surprise_score", "dxy_direction", "vix_level", "gold_momentum_5d", "time_since_last_high"]
X_raw = df[feature_cols]

pipe = Pipeline([("impute", SimpleImputer(strategy="mean")), ("scale", StandardScaler())])
X = pd.DataFrame(pipe.fit_transform(X_raw), columns=feature_cols)
print(X.head())

# correlation heatmap
corr_df = pd.concat([X_raw, df["pip_move_4h"]], axis=1)
plt.figure(figsize=(8, 6))
sns.heatmap(corr_df.corr(), annot=True, cmap="coolwarm")
plt.tight_layout()
plt.savefig("news_impact/correlation_heatmap.png")
plt.show()
```

```python
out = pd.concat([df, X.add_suffix("_scaled")], axis=1)
out.to_csv("news_impact/data/features_labeled.csv", index=False)
```

```python
import xgboost as xgb
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(df["impact_label"])

tscv = TimeSeriesSplit(n_splits=4)

for fold, (train_idx, test_idx) in enumerate(tscv.split(X)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = xgb.XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05, eval_metric="mlogloss")
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    acc = accuracy_score(y_test, preds)
    f1 = f1_score(y_test, preds, average="macro")
    print(f"Fold {fold}: acc={acc:.3f}, macro_f1={f1:.3f}")
```

```python
probs = model.predict_proba(X)
high_idx = list(le.classes_).index("HIGH")
for thresh in [0.5, 0.6]:
    pred_high = probs[:, high_idx] > thresh
    actual_high = (y == high_idx)
    tp = (pred_high & actual_high).sum()
    fp = (pred_high & ~actual_high).sum()
    fn = (~pred_high & actual_high).sum()
    precision = tp / (tp + fp) if (tp+fp) else 0
    recall = tp / (tp + fn) if (tp+fn) else 0
    print(f"thresh={thresh}: precision={precision:.2f}, recall={recall:.2f}")
```

```python
import torch
import torch.nn as nn
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import StandardScaler

df["dxy_ret"] = df["dxy"].pct_change()

feature_cols_lstm = ["surprise_score", "pip_move_4h", "vix", "dxy_ret"]

feat_scaler = StandardScaler()
target_scaler = StandardScaler()

data_scaled = feat_scaler.fit_transform(df[feature_cols_lstm].fillna(0))
target_scaled = target_scaler.fit_transform(df[["pip_move_4h"]])
event_dummies = pd.get_dummies(df["event_type"], prefix="evt").astype(float)
df = pd.concat([df, event_dummies], axis=1)

feature_cols_lstm = ["surprise_score", "pip_move_4h", "vix", "dxy_ret"]
seq_len = 10
X_seq, y_seq = [], []
for i in range(seq_len, len(data_scaled)):
    X_seq.append(data_scaled[i-seq_len:i])
    y_seq.append(target_scaled[i])

X_seq = torch.tensor(X_seq, dtype=torch.float32)
y_seq = torch.tensor(y_seq, dtype=torch.float32)

split = int(len(X_seq) * 0.8)
X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]
class PipLSTM(nn.Module):
    def __init__(self, input_size=4, hidden_size=4, num_layers=1):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.dropout = nn.Dropout(0.4)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])

model = PipLSTM()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=1e-3)
loss_fn = nn.MSELoss()

best_mae = float("inf")
best_state = None
patience = 20
no_improve = 0

for epoch in range(250):
    optimizer.zero_grad()
    pred = model(X_train)
    loss = loss_fn(pred, y_train)
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()

    model.eval()
    with torch.no_grad():
        test_pred = model(X_test)
        test_mae = mean_absolute_error(
            target_scaler.inverse_transform(y_test.numpy()),
            target_scaler.inverse_transform(test_pred.numpy())
        )
    model.train()

    if test_mae < best_mae:
        best_mae = test_mae
        best_state = model.state_dict()
        no_improve = 0
    else:
        no_improve += 1

    if epoch % 10 == 0:
        print(f"epoch {epoch}: train_loss={loss.item():.3f}, test_mae={test_mae:.2f}")

    if no_improve >= patience:
        print(f"Early stopping at epoch {epoch}")
        break

model.load_state_dict(best_state)
print(f"Best Test MAE: {best_mae:.2f}")

plt.figure(figsize=(8,4))
plt.plot(y_test.numpy(), label="actual")
plt.plot(test_pred.numpy(), label="predicted")
plt.legend()
plt.savefig("news_impact/actual_vs_predicted.png")
plt.show()

torch.save(model.state_dict(), "news_impact/pip_lstm.pt")
```
