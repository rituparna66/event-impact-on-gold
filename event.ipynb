```python
import pandas as pd
```

```python
df = pd.read_csv("all_events_final.csv")
```

```python
df.head(5)
```

```python
df.shape
```

```python
mean_abs_pip = df.groupby("event_type")["pip_move_4h"].mean().sort_values(ascending=False)
print(mean_abs_pip)
```

```python
import matplotlib.pyplot as plt

mean_abs_pip.plot(kind="bar", color=["#c0392b","#2980b9","#27ae60"])
plt.title("Mean Absolute Pip Move (4H) by Event Type")
plt.ylabel("Mean |pip move|")
plt.xlabel("Event Type")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("mean_pip_move_by_event.png")
plt.show()
```
